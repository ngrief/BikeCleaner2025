```python
import pandas as pd

def load_and_clean_data(file_paths):
    """
    Load, clean, and merge Citi Bike data from multiple files.
    
    Parameters:
    file_paths (list of str): List of file paths for the Citi Bike datasets.
    
    Returns:
    pd.DataFrame: Cleaned and merged Citi Bike data.
    """
    dataframes = []
    
    for file in file_paths:
        df = pd.read_csv(file, low_memory=False)
        
        # Standardize column names (strip spaces and lowercase)
        df.columns = df.columns.str.strip().str.lower()
        
        # Print columns for debugging
        print(f"Columns in {file}: {df.columns.tolist()}")
        
        # Convert datetime columns if present
        if 'starttime' in df.columns and 'stoptime' in df.columns:
            df['starttime'] = pd.to_datetime(df['starttime'], errors='coerce')
            df['stoptime'] = pd.to_datetime(df['stoptime'], errors='coerce')
        
        # Drop duplicates
        df.drop_duplicates(inplace=True)
        
        # Check for missing essential columns and proceed accordingly
        essential_cols = ['starttime', 'stoptime', 'start station id', 'end station id']
        existing_cols = [col for col in essential_cols if col in df.columns]
        if existing_cols:
            df.dropna(subset=existing_cols, inplace=True)
        else:
            print(f"Warning: Missing essential columns in {file}")
        
        dataframes.append(df)
    
    # Concatenate all DataFrames into one
    merged_df = pd.concat(dataframes, ignore_index=True)
    
    return merged_df

# List of file paths
file_paths = ["Resources/202501-citibike-tripdata_1.csv", "Resources/202501-citibike-tripdata_2.csv", "Resources/202501-citibike-tripdata_3.csv"]

# Load, clean, and merge data
cleaned_data = load_and_clean_data(file_paths)

# Sample 5% of the merged data
sampled_data = cleaned_data.sample(frac=0.05, random_state=42)

# Save the sampled data to a new file
sampled_data.to_csv("sampled_citibike_data.csv", index=False)

print("Data cleaning, merging, and sampling complete. Saved as 'sampled_citibike_data.csv'")
```
